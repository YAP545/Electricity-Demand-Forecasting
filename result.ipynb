```python
import pandas as pd

results = {
    "Model": ["LSTM", "RNN", "TCN", "CNN"],
    "RMSE": [0.01581, 0.01614, 0.01684, 0.01716],
    "MAE": [0.01163, 0.01211, 0.01230, 0.01283]
}

results_df = pd.DataFrame(results)

print(results_df)

# Save results
results_df.to_csv("model_comparison_results.csv", index=False)

print("Results file saved successfully!")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

results_df = pd.read_csv("model_comparison_results.csv")

# RMSE Comparison
plt.figure(figsize=(8,5))
plt.bar(results_df["Model"], results_df["RMSE"])
plt.title("RMSE Comparison of Models")
plt.xlabel("Model")
plt.ylabel("RMSE")
plt.show()

# MAE Comparison
plt.figure(figsize=(8,5))
plt.bar(results_df["Model"], results_df["MAE"])
plt.title("MAE Comparison of Models")
plt.xlabel("Model")
plt.ylabel("MAE")
plt.show()
```

```python
results_df.to_excel("Electricity_Demand_Forecasting_Results.xlsx",
                    index=False)

print("Excel report generated successfully!")
```
